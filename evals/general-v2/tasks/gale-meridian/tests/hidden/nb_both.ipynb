```r
library(data.table)
library(jsonlite)
a <- data.table(id = c(1L, 2L, 3L), score = c(2.5, 0.5, 1.0))
b <- data.table(id = c(2L, 3L, 4L), w = c(10, 20, 30))
m <- merge(a, b, by = "id")[order(id)]
cat("RCW_BOTH_OK\n")
cat(sprintf("WSUM %.1f\n", sum(m$score * m$w)))
cat(sprintf("NROW %d\n", nrow(m)))
m2 <- data.table(id = m$id, tag = c("t2", "t3"))
cat(sprintf("JSON %s\n", toJSON(m2, dataframe = "rows")))
```
