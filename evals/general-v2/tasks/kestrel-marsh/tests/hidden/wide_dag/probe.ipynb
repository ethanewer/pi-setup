# Hidden probe: wide DAG metrics


```r
library(jsonlite)
spec <- '{"nodes":["S1","S2","M1","M2","M3","O"],"edges":[["S1","M1"],["S1","M2"],["S2","M2"],["S2","M3"],["M1","O"],["M2","O"],["M3","O"]]}'
dag <- fromJSON(spec)
nodes <- as.character(dag$nodes)
edges <- as.matrix(dag$edges)
sources <- setdiff(nodes, unique(as.vector(edges[, 2])))
sinks <- setdiff(nodes, unique(as.vector(edges[, 1])))
indegM2 <- sum(as.vector(edges[, 2]) == "M2")
cat(sprintf("H3 nodes=%d edges=%d sources=%d sinks=%d indegM2=%d\n", length(nodes), nrow(edges), length(sources), length(sinks), indegM2))
```
