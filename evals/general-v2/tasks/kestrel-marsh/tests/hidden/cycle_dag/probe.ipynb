# Hidden probe: cyclic DAG + jsonlite roundtrip


```r
library(jsonlite)
spec <- '{"name":"feedback-loop","nodes":["A","B","C"],"edges":[["B","A"],["C","B"],["A","C"]]}'
dag <- fromJSON(spec)
nodes <- as.character(dag$nodes)
edges <- as.matrix(dag$edges)
sources <- setdiff(nodes, unique(as.vector(edges[, 2])))
cat(sprintf("H1 nodes=%d edges=%d sources=%d\n", length(nodes), nrow(edges), length(sources)))
```

```r
obj <- list(study = "obs-7", n = 120L, arms = c("ctl", "trt"))
back <- fromJSON(toJSON(obj, auto_unbox = TRUE))
cat(sprintf("H1 roundtrip=%s n=%d arms=%d\n", identical(back$study, obj$study), back$n, length(back$arms)))
```
