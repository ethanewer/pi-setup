```r
library(data.table)
library(jsonlite)
dt <- data.table(g = c("a", "a", "b", "b", "b"), x = c(1L, 2L, 3L, 4L, 5L))
agg <- dt[, .(total = sum(x)), by = g][order(g)]
cat("RCW_VISIBLE_OK\n")
for (i in seq_len(nrow(agg))) cat(sprintf("SUM %s %d\n", agg$g[i], agg$total[i]))
cat(sprintf("PAYLOAD %s\n", toJSON(list(units = nrow(dt), groups = nrow(agg)), auto_unbox = TRUE)))
```
